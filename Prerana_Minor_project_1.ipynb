```python
# ==========================================
# Feature 1 : Read WhatsApp Chat File
# ==========================================

from datetime import datetime
import numpy as np

with open("/content/hostel_bois.txt","r",encoding="utf-8") as file:
    content = file.read()

lines = content.split("\n")

print("Total Raw Lines :", len(lines))
```

## Feature 1 — Chat Parser

This parser reads every line, skips system messages, counts media and deleted messages, and stores all valid messages as dictionaries.

```python
# ==========================================
# Feature 1 : WhatsApp Chat Parser
# ==========================================

messages = []

system_count = 0
media_count = 0
deleted_count = 0

for line in lines:

    line = line.strip()

    # Skip empty lines
    if line == "":
        continue

    # Handle multiline messages
    if len(line) < 8 or line[2] != "/" or line[5] != "/":
        if len(messages) > 0:
            messages[-1]["text"] += " " + line
        continue

    parts = line.split(" - ",1)

    if len(parts) != 2:
        continue

    timestamp = parts[0]
    remaining = parts[1]

    # System message
    if ": " not in remaining:
        system_count += 1
        continue

    sender, text = remaining.split(": ",1)

    if text == "<Media omitted>":
        media_count += 1

    if text == "This message was deleted":
        deleted_count += 1

    messages.append({
        "timestamp": timestamp,
        "sender": sender,
        "text": text
    })

print("═"*55)
print("CHAT PARSER SUCCESS")
print("═"*55)

print("Real Messages :",len(messages))
print("System        :",system_count)
print("Media         :",media_count)
print("Deleted       :",deleted_count)
```

## Feature 2 — Group Overview

Calculate participants, message count, date range and percentage contribution of every member.

```python
# ==========================================
# Feature 2 : Group Overview
# ==========================================

member_count = {}
participants = set()

first_date = datetime.strptime(
    messages[0]["timestamp"],
    "%d/%m/%y, %H:%M"
)

last_date = datetime.strptime(
    messages[-1]["timestamp"],
    "%d/%m/%y, %H:%M"
)

for msg in messages:

    name = msg["sender"]

    participants.add(name)

    if name in member_count:
        member_count[name] += 1
    else:
        member_count[name] = 1

member_ranking = sorted(
    member_count.items(),
    key=lambda x:x[1],
    reverse=True
)

total_messages = len(messages)
total_days = (last_date.date()-first_date.date()).days+1

print("╔"+"═"*58+"╗")
print("║"+"GROUP OVERVIEW".center(58)+"║")
print("╠"+"═"*58+"╣")

print(f"║ Group        : Hostel Bois 4ever{' '*26}║")
print(f"║ Period       : {first_date.strftime('%d %b %Y')} → {last_date.strftime('%d %b %Y')}      ║")
print(f"║ Duration     : {total_days} Days{' '*34}║")
print(f"║ Messages     : {total_messages}{' '*38}║")
print(f"║ Participants : {len(participants)}{' '*39}║")

print("╠"+"═"*58+"╣")
print("║ MESSAGES PER PERSON"+" "*37+"║")
print("╟"+"─"*58+"╢")

max_msg = member_ranking[0][1]

for name,count in member_ranking:

    percent = count/total_messages*100

    bar = "█"*max(1,int((count/max_msg)*18))

    line = f" {name:<7} {bar:<18} {count:>4} ({percent:>4.1f}%)"

    print("║"+f"{line:<58}"+"║")

print("╚"+"═"*58+"╝")
```

## Feature 3 — Most Active Day & Peak Hour

Find the busiest date and busiest hour across the entire chat.

```python
# ==========================================
# Feature 3 : Activity Highlights
# ==========================================

day_count = {}
hour_count = {}

for msg in messages:

    dt = datetime.strptime(msg["timestamp"], "%d/%m/%y, %H:%M")

    day = dt.date()
    hour = dt.hour

    if day in day_count:
        day_count[day] += 1
    else:
        day_count[day] = 1

    if hour in hour_count:
        hour_count[hour] += 1
    else:
        hour_count[hour] = 1

busiest_day = max(day_count, key=day_count.get)
busiest_hour = max(hour_count, key=hour_count.get)

print("╔"+"═"*58+"╗")
print("║"+"ACTIVITY HIGHLIGHTS".center(58)+"║")
print("╠"+"═"*58+"╣")

print(f"║ 🔥 Busiest Day : {busiest_day.strftime('%d %B %Y')} ({day_count[busiest_day]} msgs)".ljust(59)+"║")
print(f"║ 🕒 Peak Hour   : {busiest_hour:02d}:00 - {busiest_hour+1:02d}:00 ({hour_count[busiest_hour]} msgs)".ljust(59)+"║")

print("╚"+"═"*58+"╝")
```

## Feature 4 — NumPy Activity Heatmap

Create a 6 × 24 NumPy matrix and render it using block characters.

```python
# ==========================================
# Feature 4 : NumPy Heatmap
# ==========================================

members = [name for name, _ in member_ranking]

heatmap = np.zeros((6,24),dtype=int)

for msg in messages:

    row = members.index(msg["sender"])

    dt = datetime.strptime(msg["timestamp"], "%d/%m/%y, %H:%M")

    col = dt.hour

    heatmap[row][col] += 1

print("ACTIVITY HEATMAP")
print("="*78)

print("       ", end="")

for h in range(24):
    print(f"{h:02d} ", end="")

print()

symbols = [".","░","▒","█"]

for i in range(6):

    print(f"{members[i]:<7}", end=" ")

    row_max = max(heatmap[i])

    for value in heatmap[i]:

        if row_max == 0:
            level = 0

        else:

            ratio = value / row_max

            if ratio == 0:
                level = 0
            elif ratio <= 0.25:
                level = 1
            elif ratio <= 0.50:
                level = 2
            else:
                level = 3

        print(symbols[level], end="  ")

    print()

print("="*78)
```

## Feature 5 — Favourite Words

Extract words, remove punctuation, ignore stop words and display the Top 10 using text bars.

```python
# ==========================================
# Feature 5 : Favourite Words
# ==========================================

stop_words = [
    "i","me","my","we","our","you","your",
    "he","she","they","them",
    "is","am","are","was","were","be",
    "the","a","an","and","or","to","of",
    "in","on","at","for","with","it","this",
    "that","have","has","had","today","about",
    "how","just","his","her","everyone","so"
]

word_count = {}

for msg in messages:

    text = msg["text"].lower()

    if text == "<media omitted>":
        continue

    if text == "this message was deleted":
        continue

    for ch in ".,!?():;\"'[]{}<>-/":
        text = text.replace(ch," ")

    words = text.split()

    for word in words:

        if len(word) < 2:
            continue

        if word in stop_words:
            continue

        if word in word_count:
            word_count[word] += 1
        else:
            word_count[word] = 1

top_words = sorted(
    word_count.items(),
    key=lambda x:x[1],
    reverse=True
)[:10]

print("╔"+"═"*58+"╗")
print("║"+"THIS GROUP'S FAVOURITE WORDS".center(58)+"║")
print("╠"+"═"*58+"╣")

max_word = top_words[0][1]

for word,count in top_words:

    bar = "█"*max(1,int((count/max_word)*18))

    line = f" {word:<10} {bar:<18} {count}"

    print("║"+f"{line:<58}"+"║")

print("╚"+"═"*58+"╝")
```

## Feature 6 — Response Time & Longest Silent Streak

Compute average response time and identify the longest consecutive silent streak for every participant.

```python
# ==========================================
# Feature 6 : Response Time & Silent Streak
# ==========================================

# ---------- Response Time ----------

reply_sum = {}
reply_count = {}

for person in participants:
    reply_sum[person] = 0
    reply_count[person] = 0

for i in range(1, len(messages)):

    prev = messages[i-1]
    curr = messages[i]

    if prev["sender"] != curr["sender"]:

        t1 = datetime.strptime(prev["timestamp"], "%d/%m/%y, %H:%M")
        t2 = datetime.strptime(curr["timestamp"], "%d/%m/%y, %H:%M")

        gap = (t2 - t1).total_seconds() / 60

        if gap >= 0:
            reply_sum[curr["sender"]] += gap
            reply_count[curr["sender"]] += 1

avg_reply = {}

for person in participants:

    if reply_count[person] == 0:
        avg_reply[person] = 0
    else:
        avg_reply[person] = round(reply_sum[person] / reply_count[person],1)


# ---------- Silent Days ----------

active_days = {}

for person in participants:
    active_days[person] = set()

for msg in messages:

    dt = datetime.strptime(msg["timestamp"], "%d/%m/%y, %H:%M")
    active_days[msg["sender"]].add(dt.date())

silent_days = {}

for person in participants:
    silent_days[person] = total_days - len(active_days[person])


# ---------- Output ----------

fastest = min(avg_reply, key=avg_reply.get)
slowest = max(avg_reply, key=avg_reply.get)

print("╔"+"═"*58+"╗")
print("║"+"RESPONSE PATTERNS".center(58)+"║")
print("╠"+"═"*58+"╣")

print(f"║ ⚡ Fastest Replier : {fastest} ({avg_reply[fastest]} min)".ljust(59)+"║")
print(f"║ 🐢 Slowest Replier : {slowest} ({avg_reply[slowest]} min)".ljust(59)+"║")

print("╠"+"═"*58+"╣")
print("║ LONGEST SILENT STREAKS"+" "*34+"║")
print("╟"+"─"*58+"╢")

for person, days in sorted(silent_days.items(), key=lambda x:x[1], reverse=True):

    line = f" {person:<8}: {days} days"

    print("║"+f"{line:<58}"+"║")

print("╚"+"═"*58+"╝")
```

## Feature 7 — Personality Archetypes

Assign exactly one personality archetype to every participant based on behavioural rules.

```python
# ==========================================
# Feature 7 : Personality Archetypes
# ==========================================

archetypes = {}

for person in participants:

    person_msgs = []

    for msg in messages:
        if msg["sender"] == person:
            person_msgs.append(msg)

    total = len(person_msgs)

    # Night Owl %
    night = 0

    for msg in person_msgs:

        dt = datetime.strptime(msg["timestamp"], "%d/%m/%y, %H:%M")

        if dt.hour >= 23 or dt.hour <= 4:
            night += 1

    night_ratio = night / total

    # Storyteller
    total_words = 0

    for msg in person_msgs:
        total_words += len(msg["text"].split())

    avg_words = total_words / total

    # Drama
    drama = 0

    for msg in person_msgs:

        text = msg["text"]

        if len(text) > 3 and (text.isupper() or text.count("!") >= 2):
            drama += 1

    drama_ratio = drama / total

    # Group Mom
    caring = 0

    keywords = [
        "sleep","eat","safe","please",
        "take care","drink water"
    ]

    for msg in person_msgs:

        txt = msg["text"].lower()

        for word in keywords:
            if word in txt:
                caring += 1

    # Spammer
    burst = 1
    best = 1

    for i in range(1, len(messages)):

        if messages[i]["sender"] == person and messages[i-1]["sender"] == person:
            burst += 1

            if burst > best:
                best = burst
        else:
            burst = 1

    # Ghost
    ghost = silent_days[person]

    # ---------- Final Rule ----------

    if ghost > 35:
        archetypes[person] = "👻 THE GHOST"

    elif night_ratio >= 0.60:
        archetypes[person] = "🌙 THE NIGHT OWL"

    elif avg_words >= 30:
        archetypes[person] = "📖 THE STORYTELLER"

    elif drama_ratio >= 0.30:
        archetypes[person] = "🎭 THE DRAMA QUEEN"

    elif caring >= 20:
        archetypes[person] = "❤️ THE GROUP MOM"

    elif best >= 4:
        archetypes[person] = "💬 THE SPAMMER"

    else:
        archetypes[person] = "😂 THE COMEDIAN"


print("╔"+"═"*58+"╗")
print("║"+"PERSONALITY ARCHETYPES".center(58)+"║")
print("╠"+"═"*58+"╣")

for person in sorted(participants):

    line = f" {person:<8} → {archetypes[person]}"

    print("║"+f"{line:<58}"+"║")

print("╚"+"═"*58+"╝")
```
